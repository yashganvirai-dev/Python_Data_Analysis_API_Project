```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import requests
import json
from datetime import datetime
```

```python
df = pd.read_csv("sales_data.csv")
display(df.head())
print("Shape:", df.shape)
print(df.dtypes)
```

```python
print("Missing values before cleaning:")
print(df.isnull().sum())

df["Customer_Rating"] = df["Customer_Rating"].fillna(df["Customer_Rating"].mean())
df = df.drop_duplicates()
df["Date"] = pd.to_datetime(df["Date"])

print("\nMissing values after cleaning:")
print(df.isnull().sum())
```

```python
display(df.describe())

total_revenue = df["Total_Sales"].sum()
total_quantity = df["Quantity"].sum()
average_rating = df["Customer_Rating"].mean()

print(f"Total Revenue: ₹{total_revenue:,.2f}")
print(f"Total Quantity Sold: {total_quantity}")
print(f"Average Customer Rating: {average_rating:.2f}")

product_sales = df.groupby("Product")["Total_Sales"].sum().sort_values(ascending=False)
category_sales = df.groupby("Category")["Total_Sales"].sum().sort_values(ascending=False)

display(product_sales)
display(category_sales)
```

```python
sns.set_theme(style="whitegrid")

plt.figure(figsize=(10,5))
product_sales.plot(kind="bar")
plt.title("Total Sales by Product")
plt.xlabel("Product")
plt.ylabel("Sales (₹)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,5))
category_sales.plot(kind="pie", autopct="%1.1f%%", startangle=90)
plt.title("Sales Distribution by Category")
plt.ylabel("")
plt.show()

plt.figure(figsize=(10,5))
sns.lineplot(data=df, x="Date", y="Total_Sales", marker="o")
plt.title("Daily Sales Trend")
plt.xlabel("Date")
plt.ylabel("Sales (₹)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(data=df, x="Quantity", y="Total_Sales", hue="Category", s=100)
plt.title("Quantity vs Total Sales")
plt.tight_layout()
plt.show()
```

```python
api_url = "https://jsonplaceholder.typicode.com/posts/1"

try:
    response = requests.get(api_url, timeout=10)
    response.raise_for_status()
    api_data = response.json()
    print("API Status Code:", response.status_code)
    print(json.dumps(api_data, indent=4))
    api_df = pd.DataFrame([api_data])
    display(api_df)
except requests.RequestException as e:
    print("API request failed:", e)
```

```python
top_product = product_sales.index[0]
top_category = category_sales.index[0]

report = f'''
PYTHON DATA ANALYSIS REPORT
===========================
Dataset: Sales Data
Records analyzed: {len(df)}

Total Revenue: ₹{total_revenue:,.2f}
Total Quantity Sold: {total_quantity}
Average Customer Rating: {average_rating:.2f}

Top Product: {top_product}
Top Product Sales: ₹{product_sales.iloc[0]:,.2f}

Top Category: {top_category}
Top Category Sales: ₹{category_sales.iloc[0]:,.2f}

Technologies: Python, NumPy, Pandas, Matplotlib, Seaborn, Requests, JSON
'''
print(report)

with open("final_report.txt", "w", encoding="utf-8") as f:
    f.write(report)
```
